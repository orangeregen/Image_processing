```python
from google.colab import files
from google.colab.patches import cv2_imshow
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
```

```python
drive.mount('/content/drive')
#path = '/content/drive/MyDrive/Image_processing/text.jpg'
```

```python
image = cv2.imread('/content/drive/MyDrive/Image_processing/text.jpg')
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

Просмотр исходного изображения

```python
plt.figure(figsize=(10, 8))
plt.gca().set_axis_off()
plt.imshow(gray_image, 'gray')
```

Простая бинаризация

```python
threshold = 100

ret, thresh1 = cv2.threshold(gray_image, threshold, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(15, 8))
plt.imshow(thresh1, 'gray')
plt.gca().set_axis_off()
plt.show()
```
